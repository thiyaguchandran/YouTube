**Read CSV**

```python
df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/ecommerce/bronze/raw_files/orders.csv")

display(df)
```

**Check the schema**

```python
df.printSchema()
```

**Check the record count**

```python
print("Number of records:", df.count())
```

**Write to the Bronze Delta table**

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("ecommerce.bronze.orders")
```

**Verify the bronze table**

```python
bronze_df = spark.table("ecommerce.bronze.orders")

display(bronze_df)
```

**Final record-count validation**

```python
print("Bronze record count:", bronze_df.count())
```
