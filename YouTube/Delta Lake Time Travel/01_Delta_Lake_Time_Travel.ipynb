```sql
%sql
DROP TABLE IF EXISTS workspace.default.ecommerce_orders_timetravel;
```

```sql
%sql

CREATE OR REPLACE TABLE workspace.default.ecommerce_orders_timetravel (
    order_id INT,
    customer_id INT,
    product_name STRING,
    quantity INT,
    amount DECIMAL(10,2),
    order_status STRING
)
USING DELTA;
```

```sql
%sql

INSERT INTO workspace.default.ecommerce_orders_timetravel
VALUES
(101, 1001, 'Laptop', 1, 750.00, 'Completed'),
(102, 1002, 'Mobile', 2, 900.00, 'Completed'),
(103, 1003, 'Headphones', 1, 80.00, 'Pending'),
(104, 1004, 'Keyboard', 2, 120.00, 'Completed'),
(105, 1005, 'Monitor', 1, 250.00, 'Pending');
```

```sql
%sql

SELECT *
FROM workspace.default.ecommerce_orders_timetravel
ORDER BY order_id;
```

```sql
%sql
DESCRIBE HISTORY workspace.default.ecommerce_orders_timetravel;
```

```sql
%sql

UPDATE workspace.default.ecommerce_orders_timetravel
SET order_status = 'Completed'
WHERE order_id = 103;
```

### **VERSION_AS_OF**

**TIMESTAMP AS OF**

```sql
-- Replace the timestamp with Version 1's
-- actual timestamp from DESCRIBE HISTORY

SELECT *
FROM workspace.default.ecommerce_orders_timetravel
TIMESTAMP AS OF '2026-10-27T19:50:43.000+00:00';
```
